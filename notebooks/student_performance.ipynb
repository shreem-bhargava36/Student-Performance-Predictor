```python
# Create the feature lists with the names used in preprocessing

categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

numerical_features = X.select_dtypes(
    exclude=["object"]
).columns.tolist()

print("Categorical features:", categorical_features)
print("Numerical features:", numerical_features)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features)
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_reg_train, y_reg_test, y_cls_train, y_cls_test = train_test_split(
    X,
    y_regression,
    y_classification,
    test_size=0.2,
    random_state=42,
    stratify=y_classification
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
print("Regression training target:", y_reg_train.shape)
print("Classification training target:", y_cls_train.shape)
```

```python
# Fit preprocessing only on training data
X_train_processed = preprocessor.fit_transform(X_train)

# Transform test data using the same preprocessing
X_test_processed = preprocessor.transform(X_test)

print("Processed training data:", X_train_processed.shape)
print("Processed testing data:", X_test_processed.shape)
```

```python
import scipy.sparse as sp

# Convert processed data to dense DataFrames
X_train_df = pd.DataFrame(
    X_train_processed.toarray() if sp.issparse(X_train_processed) else X_train_processed
)

X_test_df = pd.DataFrame(
    X_test_processed.toarray() if sp.issparse(X_test_processed) else X_test_processed
)

# Save processed data
X_train_df.to_csv("../dataset/X_train_processed.csv", index=False)
X_test_df.to_csv("../dataset/X_test_processed.csv", index=False)

y_reg_train.to_csv("../dataset/y_reg_train.csv", index=False)
y_reg_test.to_csv("../dataset/y_reg_test.csv", index=False)

y_cls_train.to_csv("../dataset/y_cls_train.csv", index=False)
y_cls_test.to_csv("../dataset/y_cls_test.csv", index=False)

print("Processed datasets saved successfully!")
```

```python
import pandas as pd

X_train = pd.read_csv("../dataset/X_train_processed.csv")
X_test = pd.read_csv("../dataset/X_test_processed.csv")

y_cls_train = pd.read_csv("../dataset/y_cls_train.csv").squeeze()
y_cls_test = pd.read_csv("../dataset/y_cls_test.csv").squeeze()

y_reg_train = pd.read_csv("../dataset/y_reg_train.csv").squeeze()
y_reg_test = pd.read_csv("../dataset/y_reg_test.csv").squeeze()

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_cls_train:", y_cls_train.shape)
print("y_cls_test:", y_cls_test.shape)

print("y_reg_train:", y_reg_train.shape)
print("y_reg_test:", y_reg_test.shape)

print("\nClassification training targets:")
print(y_cls_train.value_counts())

print("\nRegression target range:")
print(y_reg_train.min(), "to", y_reg_train.max())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Create model
lr_model = LogisticRegression(max_iter=1000, random_state=42)

# Train model
lr_model.fit(X_train, y_cls_train)

# Predict
y_pred_lr = lr_model.predict(X_test)

# Evaluation
print("Logistic Regression Results")
print("---------------------------")
print("Accuracy :", accuracy_score(y_cls_test, y_pred_lr))
print("Precision:", precision_score(y_cls_test, y_pred_lr))
print("Recall   :", recall_score(y_cls_test, y_pred_lr))
print("F1 Score :", f1_score(y_cls_test, y_pred_lr))

print("\nConfusion Matrix:")
print(confusion_matrix(y_cls_test, y_pred_lr))
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Create model
dt_model = DecisionTreeClassifier(random_state=42)

# Train model
dt_model.fit(X_train, y_cls_train)

# Predict
y_pred_dt = dt_model.predict(X_test)

# Evaluation
print("Decision Tree Results")
print("--------------------")
print("Accuracy :", accuracy_score(y_cls_test, y_pred_dt))
print("Precision:", precision_score(y_cls_test, y_pred_dt))
print("Recall   :", recall_score(y_cls_test, y_pred_dt))
print("F1 Score :", f1_score(y_cls_test, y_pred_dt))

print("\nConfusion Matrix:")
print(confusion_matrix(y_cls_test, y_pred_dt))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix

# Create model
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train model
rf_model.fit(X_train, y_cls_train)

# Predict
y_pred_rf = rf_model.predict(X_test)

# Evaluation
print("Random Forest Results")
print("---------------------")
print("Accuracy :", accuracy_score(y_cls_test, y_pred_rf))
print("Precision:", precision_score(y_cls_test, y_pred_rf))
print("Recall   :", recall_score(y_cls_test, y_pred_rf))
print("F1 Score :", f1_score(y_cls_test, y_pred_rf))

print("\nConfusion Matrix:")
print(confusion_matrix(y_cls_test, y_pred_rf))
```

```python
from sklearn.model_selection import GridSearchCV
from xgboost import XGBRegressor

xgb_tune = XGBRegressor(
    random_state=42,
    objective="reg:squarederror"
)

param_grid = {
    "n_estimators": [100, 200],
    "max_depth": [2, 3, 4],
    "learning_rate": [0.05, 0.1],
    "subsample": [0.8, 1.0]
}

grid_search = GridSearchCV(
    estimator=xgb_tune,
    param_grid=param_grid,
    cv=5,
    scoring="neg_mean_squared_error",
    n_jobs=-1
)

grid_search.fit(X_train, y_reg_train)

print("Best Parameters:")
print(grid_search.best_params_)

print("\nBest CV Score:")
print(-grid_search.best_score_)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

best_xgb_reg = grid_search.best_estimator_

y_pred_tuned = best_xgb_reg.predict(X_test)

mae = mean_absolute_error(y_reg_test, y_pred_tuned)
mse = mean_squared_error(y_reg_test, y_pred_tuned)
rmse = np.sqrt(mse)
r2 = r2_score(y_reg_test, y_pred_tuned)

print("Tuned XGBoost Regression Results")
print("--------------------------------")
print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R²  :", r2)
```
